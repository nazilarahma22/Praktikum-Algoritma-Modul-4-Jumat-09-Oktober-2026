```python
ulang = "y"

while ulang.lower() == "y":
    bulan = int(input("Masukkan bulan (1-12): "))
    tahun = int(input("Masukkan tahun: "))

    # Menentukan jumlah hari berdasarkan bulan dan tahun kabisat
    if bulan in [1, 3, 5, 7, 8, 10, 12]:
        hari = 31
    elif bulan in [4, 6, 9, 11]:
        hari = 30
    elif bulan == 2:
        # Cek tahun kabisat
        if (tahun % 400 == 0) or (tahun % 4 == 0 and tahun % 100 != 0):
            hari = 29
        else:
            hari = 28
    else:
        hari = None

    if hari:
        print(f"Jumlah hari pada bulan {bulan} tahun {tahun} adalah {hari} hari.\n")
    else:
        print("Bulan tidak valid!\n")

    # Menanyakan apakah user ingin mengulang program
    ulang = input("Apakah ingin menghitung lagi? (y/t): ")

print("Program selesai. Terima kasih!")
```
